```python
# Basic environment check
import sys
import platform

print('Python version:', sys.version.split()[0])
print('Platform:', platform.platform())
```

```python
# Optional package sanity check
packages = ['numpy', 'pandas', 'matplotlib']

for package in packages:
    try:
        __import__(package)
        print(f'{package}: available')
    except Exception as exc:
        print(f'{package}: missing ({exc})')
```

## Suggested recovery steps

1. Restart the runtime if imports fail unexpectedly.
2. Re-run the install cell for any missing package.
3. Re-open the target notebook only after the environment check passes.
