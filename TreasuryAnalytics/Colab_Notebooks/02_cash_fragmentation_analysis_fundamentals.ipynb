# Cash Fragmentation Analysis Fundamentals

This notebook introduces fragmentation analysis for treasury operations using synthetic transaction data.

## Learning goals
- Measure how fragmented cash activity is across recipients, currencies, payment types, and internal teams.
- Interpret fragmentation metrics as signals for consolidation opportunity.
- Visualize when fragmentation peaks during the operating day.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

rng = np.random.default_rng(42)
print('Treasury cash fragmentation environment ready')
```

```python
n = 1200
timestamps = pd.date_range('2025-01-01', periods=n, freq='H')
df = pd.DataFrame({
    'timestamp': timestamps,
    'recipient_bank': rng.choice(['Bank A', 'Bank B', 'Bank C', 'Bank D', 'Bank E'], size=n, p=[0.25, 0.22, 0.18, 0.18, 0.17]),
    'payment_type': rng.choice(['Wire', 'ACH', 'Card Settlement', 'Payroll', 'Tax'], size=n),
    'currency': rng.choice(['INR', 'USD', 'EUR', 'SGD'], size=n, p=[0.62, 0.18, 0.12, 0.08]),
    'department': rng.choice(['AP', 'Treasury', 'Payroll', 'Shared Services'], size=n),
    'customer_segment': rng.choice(['Enterprise', 'Mid-Market', 'SME'], size=n, p=[0.35, 0.4, 0.25]),
    'amount': np.round(rng.lognormal(mean=11.2, sigma=0.75, size=n), 2),
    'risk_score': np.round(rng.uniform(0.05, 0.95, size=n), 3)
})
df.head()
```

```python
unique_recipients = df['recipient_bank'].nunique()
unique_types = df['payment_type'].nunique()
currency_fragmentation = df['currency'].nunique()
department_fragmentation = df['department'].nunique()
fragmentation_score = (unique_recipients + unique_types + currency_fragmentation + department_fragmentation) / 4

risk_by_segment = df.groupby('customer_segment')['risk_score'].mean().round(3)
currency_analysis = df.groupby('currency').agg(amount_sum=('amount', 'sum'), transaction_count=('amount', 'count'), avg_risk=('risk_score', 'mean')).round(2)
df['hour'] = df['timestamp'].dt.hour
hourly_volume = df.groupby('hour')['amount'].sum().round(2)

print(f'Unique recipient banks: {unique_recipients}')
print(f'Payment types: {unique_types}')
print(f'Currencies involved: {currency_fragmentation}')
print(f'Departments involved: {department_fragmentation}')
print(f'Overall fragmentation score: {fragmentation_score:.2f}')

print('Risk Score by Segment:')
print(risk_by_segment)

print('Currency-Based Fragmentation Patterns:')
print(currency_analysis)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
currency_analysis['transaction_count'].plot(kind='bar', ax=axes[0], color='#0ea5e9')
axes[0].set_title('Transactions by Currency')
axes[0].set_ylabel('Count')

hourly_volume.plot(ax=axes[1], color='#1d4ed8', linewidth=2)
axes[1].set_title('Hourly Treasury Volume')
axes[1].set_ylabel('Amount')
axes[1].set_xlabel('Hour of Day')

plt.tight_layout()
plt.show()

print('Fragmentation analysis complete')
```
