# Predictive Analytics and Time Series Forecasting

This notebook uses synthetic treasury cash-flow data to demonstrate trend, seasonality, and short-horizon forecasting.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

rng = np.random.default_rng(42)
```

```python
n_samples = 365
dates = pd.date_range('2024-01-01', periods=n_samples, freq='D')
base_flow = 1_000_000 + rng.normal(0, 20_000, n_samples)
seasonal = 50_000 * np.sin(2 * np.pi * np.arange(n_samples) / 30)
trend = np.arange(n_samples) * 100
noise = rng.normal(0, 15_000, n_samples)
cash_flow = base_flow + seasonal + trend + noise
df = pd.DataFrame({'date': dates, 'cash_flow': np.round(cash_flow, 2)})
df['day_index'] = np.arange(len(df))
df['is_weekend'] = df['date'].dt.dayofweek >= 5
df.head()
```

### What this cell is doing

```python
train = df.iloc[:-30].copy()
test = df.iloc[-30:].copy()
model = LinearRegression()
model.fit(train[['day_index']], train['cash_flow'])
test['forecast'] = model.predict(test[['day_index']])
mae = mean_absolute_error(test['cash_flow'], test['forecast'])

print(f'Train rows: {len(train)} | Test rows: {len(test)}')
print(f'Mean absolute error: ₹{mae:,.0f}')
print(f'Weekend vs weekday average: ₹{df[df["is_weekend"]]["cash_flow"].mean():,.0f} vs ₹{df[~df["is_weekend"]]["cash_flow"].mean():,.0f}')
test[['date', 'cash_flow', 'forecast']].head()
```

```python
plt.figure(figsize=(13, 5))
plt.plot(df['date'], df['cash_flow'], label='Actual cash flow', color='#0ea5e9')
plt.plot(test['date'], test['forecast'], label='Forecast', color='#ef4444', linewidth=2)
plt.title('Treasury Cash Flow Forecast')
plt.xlabel('Date')
plt.ylabel('Cash Flow')
plt.legend()
plt.tight_layout()
plt.show()
```
