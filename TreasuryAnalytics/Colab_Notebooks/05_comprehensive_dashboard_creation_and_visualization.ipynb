# Comprehensive Dashboard Creation and Visualization

This notebook creates a teaching dashboard from synthetic treasury metrics so students can connect liquidity, portfolio value, risk, and compliance on one page.


```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go

rng = np.random.default_rng(42)
```

```python
def generate_dashboard_data(n_periods=120):
    dates = pd.date_range('2025-01-01', periods=n_periods, freq='D')
    dashboard_df = pd.DataFrame({
        'date': dates,
        'total_cash_flow': rng.normal(9_500_000, 1_200_000, n_periods).cumsum() / 3,
        'portfolio_value': rng.normal(250_000_000, 7_000_000, n_periods).cumsum() / 8,
        'overall_risk_score': np.clip(rng.normal(0.48, 0.12, n_periods), 0.05, 0.95),
        'compliance_score': np.clip(rng.normal(0.88, 0.05, n_periods), 0.6, 0.99),
    })
    return dashboard_df

dashboard_df = generate_dashboard_data()
print(f'Average compliance score: {dashboard_df["compliance_score"].mean():.2f}')
dashboard_df.head()
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=dashboard_df['date'], y=dashboard_df['total_cash_flow'], name='Total Cash Flow'))
fig.add_trace(go.Scatter(x=dashboard_df['date'], y=dashboard_df['portfolio_value'], name='Portfolio Value', yaxis='y2'))
fig.add_trace(go.Scatter(x=dashboard_df['date'], y=dashboard_df['overall_risk_score'], name='Risk Score', yaxis='y3'))
fig.add_trace(go.Scatter(x=dashboard_df['date'], y=dashboard_df['compliance_score'], name='Compliance Score', yaxis='y4'))
fig.update_layout(
    title='Treasury Analytics Comprehensive Dashboard',
    xaxis_title='Date',
    yaxis=dict(title='Cash Flow'),
    yaxis2=dict(title='Portfolio Value', overlaying='y', side='right'),
    yaxis3=dict(title='Risk Score', overlaying='y', side='left', position=0.05),
    yaxis4=dict(title='Compliance Score', overlaying='y', side='right', position=0.95),
    hovermode='x unified',
    height=700,
)
fig.show()
```

```python
summary_metrics = {
    'total_cash_flow': dashboard_df['total_cash_flow'].sum(),
    'avg_portfolio_value': dashboard_df['portfolio_value'].mean(),
    'avg_risk_score': dashboard_df['overall_risk_score'].mean(),
    'overall_compliance': dashboard_df['compliance_score'].mean(),
}

print('Key Performance Indicators:')
for metric, value in summary_metrics.items():
    print(metric, round(value, 2))
```
