# Basic Setup and Data Preprocessing
## Overview
**Level:** Beginner
**Time Estimate:** 30 minutes
**Description:** Complete Colab environment setup and initial data preparation for Treasury Analytics

This notebook provides comprehensive coverage of Treasury Analytics techniques using Google Colab.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print('🏦 Treasury Analytics - Data Preprocessing Setup')
print('✅ Google Colab environment ready for Treasury Analytics')

# Sample data for demonstration
np.random.seed(42)

# Create synthetic treasury data
def generate_sample_data(n_samples=1000):
    data = {
        'transaction_id': range(1, n_samples + 1),
        'amount': np.random.lognormal(8, 1.5, n_samples),
        'customer_segment': np.random.choice(['premium', 'standard', 'basic'], n_samples),
        'risk_score': np.random.beta(1, 3, n_samples),
        'timestamp': pd.date_range('2024-01-01', periods=n_samples, freq='H'),
        'is_anomaly': np.random.randint(0, 2, n_samples)
    }
    return pd.DataFrame(data)

# Load and explore data
sample_data = generate_sample_data(500)

print(f'📊 Dataset loaded: {len(sample_data)} transactions')
print(f'💰 Total transaction volume: ₹{sample_data["amount"].sum():,.0f}')
print(f'🚨 Anomaly rate: {sample_data["is_anomaly"].mean():.1%}')

# Basic data exploration
print('\n📋 Data Summary:')
print(sample_data.describe())
```
