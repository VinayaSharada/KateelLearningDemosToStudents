# Advanced Features and Model Deployment

This notebook simulates a treasury model registry and shows how teams can compare performance, compliance, and deployment readiness before promoting a model.


```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

rng = np.random.default_rng(42)
```

```python
dates = pd.date_range('2025-01-01', periods=180, freq='D')
versions = rng.choice(['v1.0', 'v1.1', 'v2.0', 'v2.1'], size=len(dates))
statuses = rng.choice(['active', 'testing', 'deprecated'], size=len(dates), p=[0.5, 0.35, 0.15])
df = pd.DataFrame({
    'date': dates,
    'model_version': versions,
    'deployment_status': statuses,
    'performance_score': np.clip(rng.normal(0.83, 0.06, len(dates)), 0.55, 0.98),
    'compliance_score': np.clip(rng.normal(0.9, 0.04, len(dates)), 0.7, 0.99),
    'liquidity_ratio': np.clip(rng.normal(0.78, 0.08, len(dates)), 0.4, 0.98),
    'risk_score': np.clip(rng.normal(0.42, 0.1, len(dates)), 0.1, 0.85),
})
df['deployment_readiness'] = df['performance_score'] * 0.4 + df['compliance_score'] * 0.3 + df['liquidity_ratio'] * 0.3
df.head()
```

```python
model_performance = df.groupby('model_version').agg(
    performance_mean=('performance_score', 'mean'),
    performance_std=('performance_score', 'std'),
    active_days=('deployment_status', lambda x: (x == 'active').sum()),
).round(3)
deployment_analysis = df['deployment_status'].value_counts().to_dict()

print('Model Performance Overview:')
print(model_performance)
print('\nDeployment Status Counts:')
print(deployment_analysis)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
df.groupby('model_version')['performance_score'].mean().plot(kind='bar', ax=axes[0,0], color='skyblue')
axes[0,0].set_title('Performance by Model Version')
df['deployment_status'].value_counts().plot(kind='pie', ax=axes[0,1], autopct='%1.1f%%')
axes[0,1].set_ylabel('')
axes[0,1].set_title('Deployment Status Distribution')
df = df.sort_values('date')
df['deployment_readiness'].rolling(window=30).mean().plot(ax=axes[1,0], color='green')
axes[1,0].set_title('30-day Deployment Readiness')
sns.heatmap(df[['performance_score', 'compliance_score', 'liquidity_ratio', 'risk_score']].corr(), annot=True, cmap='coolwarm', ax=axes[1,1])
axes[1,1].set_title('Metric Correlations')
plt.tight_layout()
plt.show()
```
