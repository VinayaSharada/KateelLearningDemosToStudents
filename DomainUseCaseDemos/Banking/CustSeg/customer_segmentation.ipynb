# Bank Customer Segmentation Analysis

This notebook demonstrates customer segmentation for banking using synthetic customer data, K-Means clustering, and RFM-style business interpretation.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

rng = np.random.default_rng(42)
```

```python
n = 1500
df = pd.DataFrame({
    'age': rng.integers(21, 70, size=n),
    'annual_income': rng.normal(950_000, 420_000, size=n).clip(150_000, 3_000_000),
    'account_balance': rng.normal(420_000, 260_000, size=n).clip(10_000, 2_500_000),
    'monthly_txn_count': rng.integers(2, 45, size=n),
    'credit_score': rng.normal(690, 60, size=n).clip(450, 850),
    'days_since_last_txn': rng.integers(1, 180, size=n),
    'is_churned': rng.choice([0, 1], size=n, p=[0.86, 0.14]),
})
df.head()
```

```python
features = ['annual_income', 'account_balance', 'monthly_txn_count', 'credit_score', 'days_since_last_txn']
scaler = StandardScaler()
X = scaler.fit_transform(df[features])
model = KMeans(n_clusters=4, random_state=42, n_init=10)
df['cluster'] = model.fit_predict(X)
cluster_summary = df.groupby('cluster')[features + ['is_churned']].mean().round(2)
cluster_summary
```

```python
rfm_score = (
    (1 / (1 + df['days_since_last_txn'])) * 0.35
    + (df['monthly_txn_count'] / df['monthly_txn_count'].max()) * 0.3
    + (df['account_balance'] / df['account_balance'].max()) * 0.35
)
df['rfm_score'] = rfm_score.round(3)
df['rfm_segment'] = pd.qcut(df['rfm_score'], q=4, labels=['Low Value', 'Developing', 'Loyal', 'Premier'])
df[['cluster', 'rfm_segment', 'annual_income', 'account_balance', 'monthly_txn_count', 'credit_score']].head()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
scatter = axes[0].scatter(df['annual_income'], df['account_balance'], c=df['cluster'], cmap='viridis', alpha=0.5)
axes[0].set_title('Income vs Balance by Cluster')
axes[0].set_xlabel('Annual Income')
axes[0].set_ylabel('Account Balance')
plt.colorbar(scatter, ax=axes[0])
df['rfm_segment'].value_counts().plot(kind='bar', ax=axes[1], color='#60a5fa')
axes[1].set_title('RFM Segment Distribution')
axes[1].set_xlabel('Segment')
axes[1].set_ylabel('Customers')
plt.tight_layout()
plt.show()
```
